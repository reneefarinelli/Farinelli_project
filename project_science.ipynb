**First task**: 
Extract all students that study Online, have an attendance_rate smaller than 50, and have more than 5 missed_deadlines. Store them in a dataframe called at_risk.

First, I have to filter the DataFrame to select only the specified students, and I can do that using the comparison operators (==, >, and <). Then I can combine these conditions with the operator &, which represents “and” in pandas.

```python
at_risk = students[(students['study_mode'] == 'Online') & (students['attendance_rate']<50) 
                   & (students['missed_deadlines'] > 5)]
```

**Second task**:
Then compute, for each course_type, how many of those students there are and what fraction of them has dropped out. Store the result in a dataframe called at_risk_by_course, with columns students and dropout_rate, sorted by decreasing dropout_rate.

For this task, first of all I use pandas’ groupby() method to group row (so the students) by course type.

```python
by_course = at_risk.groupby('course_type')
```

I can create a DataFrame containing each course type and the number of students enrolled in it. 
In order to do this, after grouping the data by course type, I can use size() to count the rows in each group. 
Since each row represents one student, in fact, I can use these counts to highlight how many students there are. 
So, in this way I can see how the selected students are distributed across course types.

```python
at_risk_by_course = pd.DataFrame({"students": by_course.size()})
at_risk_by_course
```

At this point, I can sum the dropout indicator for each course type in order to count how many students dropped out. 
I can do this because students who dropped out are coded as 1, and all other students are coded as 0, so only students who dropped out contribute to the sum.

I can then calculate the dropout rate by dividing the number of dropouts by the total number of students in each course type.

```python
dropouts = by_course['dropout'].sum()
at_risk_by_course['dropout_rate'] = (dropouts / at_risk_by_course['students'])
```

Now I sort the DataFrame by dropout rate in descending order. To do this, I use the sort_values() method and set ascending=False.

```python
at_risk_by_course = at_risk_by_course.sort_values('dropout_rate', ascending=False)
at_risk_by_course
```

**Third task**: 
Add to the dataframe the following columns: 
- pass_rate: the ratio between units_passed and units_enrolled
- gpa_change: the difference between gpa_semester_2 and gpa_semester_1
- gpa_trend: a string that is "improving" if gpa_change is larger than 0.5, "declining" if gpa_change is smaller than -0.5, and   "stable" otherwise.

I start by creating the first two columns required for the task.

```python
students['pass_rate'] = students['units_passed']/ students['units_enrolled']
students['gpa_change'] = students['gpa_semester_2'] - students['gpa_semester_1']
```

Now I need to create a gpa_trend column whose values depend on specific conditions. 
In order to do this, I define a function with a series of if, elif, and else statements. 

I then use apply() with axis=1 to run the function on each row of the DataFrame and create the column. 
In the same line I convert the column to the "string" data type, as required by the task.

```python
def trend(row):
    if row['gpa_change'] > 0.5:
        return 'improving'
    elif row['gpa_change'] < - 0.5:
            return 'declining'
    else:
        return 'stable'
    
students['gpa_trend'] = students.apply(trend, axis=1).astype("string")
students['gpa_trend'].dtype
```

Now I can display the values in the new columns for the first 15 rows of the DataFrame to check that they look as expected.

```python
students[['pass_rate' , 'gpa_change' , 'gpa_trend']].head(15)
```

The classification works as it should: students with gpa_change > 0.5 are labeled improving, those with gpa_change < 0.5 are labeled declining, and the remaining students are labeled stable.

**Fifth task**: 
Compute the dropout rate for each value of gpa_trend, and the person_id of the 10 students with the largest GPA drop.

For the first part of the exercise, as in the second task, I use groupby() to create a summary DataFrame, this time grouping the rows by gpa_trend. 
Then I calculate the dropout rate in the same way by summing the dropout indicator for each group and dividing the result by the total number of students in that group (which is calculated using the size() method).

```python
a = students.groupby('gpa_trend')
a1 = pd.DataFrame({'students': a.size()})
a1
```

Now I can calculate the dropout_rate.

```python
dropouts = a['dropout'].sum()
a1['dropout_rate'] = (dropouts / a1['students'])
a1
```

Now, to answer the second part of the question, I use gpa_change, which I calculated by subtracting the first-semester gpa from the second-semester gpa. A negative value (gpa_change < 0) means that the student’s GPA decreased.

So, in order to find the 10 students with the largest gpa drops, I can keep the rows with a negative gpa_change and sort them by that column in ascending order, which is the default for sort_values(). 
This puts the most negative values first, so that I can take the first 10 rows.

```python
top_10_GPA_drop = students.sort_values('gpa_change').head(10)
top_10_GPA_drop[['person_id' , 'gpa_change']]
```
