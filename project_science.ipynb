**TASK NUMBER TWO**:
Then compute, for each course_type, how many of those students there are and what fraction of them has dropped out. Store the result in a dataframe called at_risk_by_course, with columns students and dropout_rate, sorted by decreasing dropout_rate.

For this task, first of all we use pandas’ groupby( ) method to group row (so the students) by course type.

```python
by_course = at_risk.groupby('course_type')
```

We can create a DataFrame containing each course type and the number of students enrolled in it. 
In order to do this, after grouping the data by course type, we can use size( ) to count the rows in each group. 
Since each row represents one student, in fact, we can use these counts to highlight how many students there are. 
So, in this way we can see how the selected students are distributed across course types.

```python
at_risk_by_course = pd.DataFrame({'students': by_course.size()})
at_risk_by_course
```

At this point, we can sum the dropout indicator for each course type in order to count how many students dropped out. 
We can do this because students who dropped out are coded as 1, and all other students are coded as 0, so only students who dropped out contribute to the sum.

We can then calculate the dropout rate by dividing the number of dropouts by the total number of students in each course type.

```python
dropouts = by_course['dropout'].sum()
at_risk_by_course['dropout_rate'] = (dropouts / at_risk_by_course['students'])
```

Now we sort the DataFrame by dropout rate in descending order. To do this, we use the sort_values( ) method and set ascending=False.

```python
at_risk_by_course = at_risk_by_course.sort_values('dropout_rate', ascending=False)
at_risk_by_course
```

**TASK NUMBER THREE**: 
Add to the dataframe the following columns: 
- pass_rate: the ratio between units_passed and units_enrolled
- gpa_change: the difference between gpa_semester_2 and gpa_semester_1
- gpa_trend: a string that is "improving" if gpa_change is larger than 0.5, "declining" if gpa_change is smaller than -0.5, and   "stable" otherwise.

We can start by creating the first two columns required for the task.

```python
students['pass_rate'] = students['units_passed']/ students['units_enrolled']
students['gpa_change'] = students['gpa_semester_2'] - students['gpa_semester_1']
```

Now we have to create a gpa_trend column whose values depend on specific conditions. 
In order to do this, we define a function with a series of "if", "elif", and "else" statements. 

Then we use apply( ) with axis=1 to run the function on each row of the DataFrame and create the column. 
In the same line we convert the column to the "string" data type, as required by the task.

```python
def trend(row):
    if row['gpa_change'] > 0.5:
        return 'improving'
    elif row['gpa_change'] < - 0.5:
            return 'declining'
    else:
        return 'stable'
    
students['gpa_trend'] = students.apply(trend, axis=1).astype("string")
students['gpa_trend'].dtype
```

Now we can display the values in the new columns for the first 15 rows of the DataFrame to check that they look as expected.

```python
students[['pass_rate' , 'gpa_change' , 'gpa_trend']].head(15)
```

The classification works as it should: students with gpa_change > 0.5 are labeled improving, those with gpa_change < 0.5 are labeled declining, and the remaining students are labeled stable.

**TASK NUMBER SIX**: 
Partition the students into the following groups, according to weekly_work_hours: 0, 1-10, 11-20, 21-30, more than 30 (all intervals include both extremes).

As we did in Exercise 3, we can define a function that uses a series of conditions (through == , < , and >) to determine which group each student belongs to. 
Then we use apply( ) to run the function on each row and store the results in the new work_hours_group column.

We check the cut-offs from lowest to highest because, doing this, we can use "<=" for every expression (except for the first and the last ones). In fact, Python evaluates each "if" or "elif" condition only if all the conditions before are false.

```python
def hours(row): 
    if row['weekly_work_hours'] == 0:
        return '0'
    elif row['weekly_work_hours'] <= 10:
        return '1-10'
    elif row['weekly_work_hours'] <= 20:
        return '11-20'
    elif row['weekly_work_hours'] <= 30: 
        return '21-30'
    else: 
        return 'more than 30'
    
students['work_hours_group'] = students.apply(hours, axis=1).astype("string")
students[['work_hours_group', 'weekly_work_hours']]
```

**TASK NUMBER SEVEN**:
Build a risk score by summing:
- 2 points if the tuition is not up to date
- 1 point if stress_level is at least 8
- 1 point if sleep_hours is smaller than 5
- 1 point if the student has a part-time job with more than 20 weekly work hours
- 2 points if attendance_rate is smaller than 60
- 1 point if missed_deadlines is larger than 3.
For each value of the risk score, compute the number of students and the dropout rate.

In order to calculate the risk score, we can define a function that starts with a score of zero and adds points for each condition that is met. 
We use independent "if" statements so that multiple conditions can contribute to the same student’s score, and we also use the operators ==, <, and > according to the scoring rules specified in the task. 
Then, in the end, the function returns the final risk score.

Then we use apply( ) and axis=1 to run the function on each row of the students DataFrame and store each student’s score in a new column called "risk_score".

Finally, we print the first 10 rows, showing the new column and the columns used to calculate it, so that we can check that the risk scores follow the scoring rules.

```python
def risk(row): 
    score = 0
    if row['tuition_up_to_date'] == 0:
        score += 2
    if row['stress_level'] >= 8: 
        score += 1
    if row['sleep_hours'] < 5: 
        score += 1
    if row['part_time_job'] == 1 and row['weekly_work_hours'] > 20: 
        score += 1
    if row['attendance_rate'] < 60: 
        score += 2
    if row['missed_deadlines'] > 3:
        score += 1
    return score

students['risk_score'] = students.apply(risk, axis=1)

students[[ 'tuition_up_to_date', 'stress_level', 'sleep_hours', 'part_time_job', 
          'weekly_work_hours', 'attendance_rate', 'missed_deadlines' , 'risk_score']].head(10)
```

Now, to answer the second part of the exercise, we can follow the same approach as in Exercises 2 and Exercise 5. 
This time, we use groupby( ) to group the rows by risk score, and then we use size( ) to count the students in each group and store these numbers in a column called "students".

Then, like we did in the other exercises, in order to calculate the dropout rate for each risk score, we sum the dropout indicator within each group to count how many students dropped out. Then we divide that count by the total number of students in the same group.

```python
risk1 = students.groupby('risk_score')
risk_table = pd.DataFrame({'students': risk1.size()})
risk_table
```

```python
dropouts = risk1['dropout'].sum()
risk_table['dropout_rate'] = (dropouts / risk_table['students'])
risk_table
```

In this way, for each value of the risk score, we can see the number of students and the dropout rate.

**TASK NUMBER NINE**: 
Join the GPD data with the students data. Compute the dropout rate for each continent.    

We can use merge( ) in pandas to join two DataFrames. 
Before merging them, we list the column names in both DataFrames to identify which column to use to make the join.

**N.B.:** I have used the cleaned GDP DataFrame (gdp_clean1) because it contains only the most recent record for each country. Otherwise, students could match multiple records for their country and could be counted more than once when calculating the dropout rate for each continent.

```python
students.columns
```

```python
gdp_clean1.columns
```

We can use the column "nationality" from students and the column "entity" from the gdp DataFrame.
Let's compare the two columns to see which countries appear in both and whether their names are written in the same way.

```python
students['nationality'].unique()
```

```python
gdp_clean1['entity'].unique()
```

We can observe that, in the students DataFrame, the United Kingdom and the United States are listed as "UK" and "USA". 
Therefore, we can use the replace( ) method to change these values to "United Kingdom" and "United States", so that they match the country names in the column "entity" of the gdp DataFrame.

```python
students['nationality'] = students['nationality'].replace({ 'USA': 'United States' , 'UK': 'United Kingdom'})
students['nationality'].unique()
```

At this point, we can use merge( ) to join the two DataFrames using the "nationality" and "entity" columns. 
In particular, we use an inner join (which is made by default) so that the result includes only countries that appear in gdp and are the country of origin of at least one student in the students DataFrame.

```python
students_gdp = pd.merge(students, gdp_clean1, left_on = 'nationality', right_on = 'entity')
students_gdp
```

Now, to calculate the dropout rate for each continent, we can follow the same approach as in the previous exercises. 
This time, we use groupby( ) to group the rows in the merged DataFrame by each student’s continent of origin. 
We then sum the dropout indicator for each continent and divide that sum by the total number of students in the same group (obtained through the size( ) method).

The column containing informations about the continent is "owid_region"

```python
students_gdp['owid_region'].unique()
```

```python
continent = students_gdp.groupby('owid_region')
continent_table = pd.DataFrame({'students': continent.size()})
continent_table
```

Now we sum the dropouts and compute the dropout_rate for each continent

```python
dropouts = continent['dropout'].sum()

continent_table['dropout_rate'] = (dropouts / continent_table['students'])
continent_table
```
